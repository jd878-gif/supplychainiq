```python
from pathlib import Path
import numpy as np, pandas as pd
from supplychainiq.generator.config import load_config
from supplychainiq.generator.load_m5 import load_m5

ROOT = Path(r"C:\Users\jeetd\Downloads\Databricks_proj\supplychainiq\supplychainiq")

cfg = load_config()
data = load_m5(cfg, ROOT)
arr = np.load(ROOT / "data/generated/daily_arrays.npz")
lost, demand = arr["lost"], arr["demand"]

# 1. New items: no history but demand in the window
new = data.hist_mean == 0
print("new-item series:", new.sum(), "| share of lost units:", round(lost[:, new].sum() / lost.sum(), 3))

# 2. Trend: window demand vs history demand
print("demand growth (window mean / history mean):",
      round(demand.mean() / data.hist_mean.mean(), 3))

# 3. Getting worse over time? Lost share per quarter
q = pd.Series(lost.sum(axis=1) / np.maximum(demand.sum(axis=1), 1), index=data.dates)
print(q.resample("QE").mean().round(3))
```
